```python
import pandas as pd
import matplotlib.pyplot as plt
from 
```

```python
df = pd.read_csv('energy-and-mining_ken.csv')
df
```

```python
rural_data = df[df['Indicator Name'] == 'Access to electricity, rural (% of rural population)']
rural_data
```

```python
urban_data = df[df['Indicator Name'] == 'Access to electricity, urban (% of urban population)']
urban_data
```

```python
rural_data = rural_data[['Year', 'Value']].rename(columns={'Value':'% Rural Access'})
rural_data
```

```python
urban_data = urban_data[['Year', 'Value']].rename(columns = {'Value' : '% Urban Access'})
urban_data
```

```python
compare_df = pd.merge(rural_data, urban_data, on='Year')
compare_df
```

```python
plt.figure()
```

```python
plt.plot(compare_df['Year'], compare_df['% Rural Access'], label = 'Rural Access %', marker='o')
```

```python
plt.plot(compare_df['Year'], compare_df['% Urban Access'], label = 'Urban Access %', marker='o')
```

```python
plt.plot(compare_df['Year'], compare_df['% Rural Access'], label = 'Rural Access %', marker = 'o')
plt.plot(compare_df['Year'], compare_df['% Urban Access'], label = 'Urban Access %', marker = 'o')

plt.title('Kenya Electrification gap between 1990-2025')
plt.xlabel('Year')
plt.ylabel('% Access')
plt.legend()
plt.grid(True)
plt.show()
```

```python
compare_df['Gap'] = compare_df['% Urban Access']- compare_df['% Rural Access']
```

```python
compare_df
```

```python
gap_df = compare_df[['Year', 'Gap']]
gap_df
```

```python
plt.plot(gap_df['Year'], gap_df['Gap'], label='% Gap', marker='o')
plt.title('Access % Gaps from 1990-2025')
plt.xlabel('Year')
plt.ylabel('% Gap')
plt.legend()
plt.grid(True)
plt.show()
```

```python
gap_df.loc[[gap_df['Gap'].idxmax()]]
```

```python
gap_df.loc[[gap_df['Gap'].idxmin()]]
```

Insights
-   The Electrification gap is highest in 2009 and lowest in 2023
-    Overall the gap shows a gradual decline over time from its peak in 2009.
-   The declining trend may be associated with increased investments in rural electrification and expanded energy access initiatives during this period, including government programs supported by independent power producers (IPPs) and development partners such as World Bank  source

```python
outage_df = df[df['Indicator Name'] == 'Value lost due to electrical outages (% of sales for affected firms)'][['Year', 'Value']].rename(columns={'Value':'Value lost'})
outage_df
```

```python
connect_df = df[df['Indicator Name'] == 'Time to obtain an electrical connection (days)'][['Year', 'Value']].rename(columns={'Value':'Time to connect'})
connect_df
```

```python
business_df = pd.merge(outage_df, connect_df, on='Year')
business_df
```

```python
plt.figure()
```

```python


scatter = plt.scatter(business_df['Time to connect'], business_df['Value lost'], c=business_df['Year'], cmap='viridis', s=100, alpha=0.5)
plt.title('outages vs connection delays')
plt.xlabel('connection days')
plt.ylabel('outages (% of sale)')
plt.grid(True, alpha=0.3)
plt.show
```

In this correlation there is a clear shift, we can see that the darkest points are older years and most lighest are recent years , revealing that the hidden tax on Kenya private sector in the form of connection delays vs outages losses has reduced over time, as evidence with earlier improvements observed in the increase energy adoptations like renewables ,suggesting that an expansion to more diverse and stable energy mix is tranlating to operational gains
